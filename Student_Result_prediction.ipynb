```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
```

```python
df=pd.read_csv(r"D:\Disk G\ML Supervise Learning Project\student_performance.csv")
```

```python
df.drop_duplicates(inplace=True)
```

```python
df
```

```python
df[df["previous_score"]<=50]
```

```python
x=df[["study_hours","attendance","assignments","previous_score"]]
y=df["final_result"]
X_train,X_test,Y_train,Y_test=train_test_split(x,y,test_size=0.2,random_state=30)
```

```python
model=LogisticRegression()
```

```python
model.fit(X_train,Y_train)
```

```python
prediction=model.predict(X_test)
```

```python
accs=accuracy_score(Y_test,prediction)
ps=precision_score(Y_test,prediction)
rs=recall_score(Y_test,prediction)
fs=f1_score(Y_test,prediction)
```
